**Task 1: Vectors**

•	Create a vector containing the ages of 6 people of your choice.

•	Print the vector's shape, dimensions, and data type.

•	Create a second vector of the same size representing "years until retirement" (65 − age) without using a loop.

•	Find and print the average age using a NumPy function (do not calculate it manually).

```python
import numpy as np

ages = np.array([18, 25, 32, 45, 50, 61])
print("===Ages Vector===\n", ages)

print("\n===Vector Information===")
print("Shape:", ages.shape)
print("Dimensions:", ages.ndim)
print("Data Type:", ages.dtype)

years_to_retirement = 65 - ages
print("\n===Years to Retirement===\n", years_to_retirement)

average_age = np.mean(ages)
print("\n===Average Age===")
print("Average Age:", average_age)
```

**Task 3: Addition & Subtraction**

•	Create two 3×3 matrices representing the marks of 3 students in 3 subjects for Term 1 and Term 2.

•	Calculate the total combined marks (Term 1 + Term 2) for each student in each subject.

•	Calculate the improvement (Term 2 − Term 1) for each student in each subject.

•	Using the improvement matrix, identify (through code, not by manually reading) which student had the highest total improvement across all subjects.

```python
import numpy as np

term1 = np.array([
    [75, 80, 85],
    [90, 85, 88],
    [60, 70, 65]
])

term2 = np.array([
    [80, 82, 89],
    [88, 87, 92],
    [75, 80, 85]
])

print("=== Term 1 Marks ===\n", term1)
print("\n=== Term 2 Marks ===\n", term2)

total_marks = term1 + term2
print("\n=== Total Combined Marks ===\n", total_marks)

improvement = term2 - term1
print("\n=== Improvement ===\n", improvement)

student_improvements = np.sum(improvement, axis=1)
best_student_index = np.argmax(student_improvements)

print("\n=== Total Improvement Per Student ===")
for i, imp in enumerate(student_improvements):
    print(f"Student {i+1}: {imp} marks")

print(f"\nStudent with the highest total improvement: Student {best_student_index + 1}")
```

**Task 4: Dot Product**

•	Create a vector of 4 item prices and a vector of the quantities purchased for each item.

•	Calculate the total bill using the dot product (not by multiplying and adding manually).

•	Create a 2×4 matrix representing quantities purchased by 2 different customers, then calculate the total bill for both customers using a single dot product operation.

•	Explain (in a comment in your code) why the shapes of the two arrays had to match a specific way for the dot product to work.

```python
import numpy as np

prices = np.array([10, 20, 5, 50])
quantities = np.array([3, 1, 10, 2])

print("Prices:", prices)
print("Quantities:", quantities)

total_bill = np.dot(prices, quantities)
print("Total bill (single customer):", total_bill)

customer_quantities = np.array([
    [3, 1, 10, 2],
    [1, 5, 2, 4]
])

customer_bills = np.dot(customer_quantities, prices)
print("\nCustomer Quantities Matrix:\n", customer_quantities)
print("Total bills for both customers:", customer_bills)
```

**Task 7: Slicing**

•	Create a 6×6 matrix filled with numbers from 1 to 36 (in order).

•	Extract the top-left 3×3 sub-matrix using slicing.

•	Extract every alternate row and every alternate column from the full matrix using step slicing.

•	Extract the last 2 rows and last 2 columns together as a single sub-matrix.

```python
import numpy as np

matrix_6x6 = np.arange(1, 37).reshape(6, 6)
print("=== Original 6x6 Matrix ===\n", matrix_6x6)

top_left_3x3 = matrix_6x6[0:3, 0:3]
print("\n=== Top-Left 3x3 Sub-Matrix ===\n", top_left_3x3)

alternating = matrix_6x6[::2, ::2]
print("\n=== Every Alternate Row & Column ===\n", alternating)

last_2x2 = matrix_6x6[-2:, -2:]
print("\n=== Last 2 Rows & 2 Columns ===\n", last_2x2)
```

**Task 9: Transpose**

•	Create a 3×5 matrix representing 3 employees' performance scores across 5 months.

•	Transpose the matrix and print its new shape.
•	Verify programmatically (using code, not by manual inspection) that transposing the matrix twice gives back the original matrix.

•	Create a 4×2 and a 3×2 matrix, and use transpose on one of them so that a dot product between the two becomes possible; then perform the dot product.

```python
import numpy as np

scores = np.array([
    [85, 90, 88, 92, 95],
    [70, 75, 80, 78, 85],
    [95, 98, 92, 90, 89]
])
print("=== Original Scores Matrix (3x5) ===\n", scores)

scores_transposed = scores.T
print("\n=== Transposed Scores Matrix (5x3) ===\n", scores_transposed)
print("New Shape:", scores_transposed.shape)

double_transpose = scores.T.T
is_identical = np.array_equal(scores, double_transpose)
print("\nDoes transposing twice give back the original matrix?", is_identical)

matrix_A = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

matrix_B = np.array([
    [9, 10],
    [11, 12],
    [13, 14]
])

matrix_B_T = matrix_B.T
dot_product_result = np.dot(matrix_A, matrix_B_T)

print("\n=== Matrix A (4x2) ===\n", matrix_A)
print("\n=== Matrix B Transposed (2x3) ===\n", matrix_B_T)
print("\n=== Dot Product Result (4x3) ===\n", dot_product_result)
```
